## PyTorch Datasets and Dataloaders

For more information, go to: https://pytorch.org/docs/stable/data.html

A `Dataset` gives access to individual examples, a `DataLoader` groups them into minibatches. For tensors that are already in memory, `torch.utils.data.TensorDataset` does this for you.

Here, we're creating custom dataset classes, which is a common practice when dealing with unique data that can't be directly used with pre-existing dataset classes provided by PyTorch.

```python
from torch.utils.data import Dataset, DataLoader
```

```python
num_features = 10
num_train_examples = 100
num_test_examples = 50
num_classes = 4

train_X = torch.rand(num_train_examples, num_features)
train_y = torch.randint(low=0, high=num_classes, size=(num_train_examples, 1))
test_X = torch.rand(num_test_examples, num_features)
test_y = torch.randint(low=0, high=num_classes, size=(num_test_examples, 1))
```

```python
print(train_X.shape)
print(test_X.shape)
print(train_y.shape)
print(test_y.shape)
```

```python
# Create a custom class MyDataset (inheriting from Dataset).
# A Dataset needs two methods: __len__ (the number of examples) and __getitem__ (the i-th example).
# The DataLoader uses them to pick examples (in random order if shuffle=True) and to stack them into batches.

class MyDataset(Dataset):
    def __init__(self, data_X, data_y):
        assert len(data_X) == len(data_y)
        self.data_X = data_X
        self.data_y = data_y

    def __len__(self):
        return len(self.data_X)

    def __getitem__(self, i):
        return self.data_X[i], self.data_y[i]
```

```python
# Example data
sample_X = torch.tensor([[1.0], [2.0], [3.0]])
sample_y = torch.tensor([0, 1, 0])

# Create the dataset
dataset = MyDataset(sample_X, sample_y)
print('Number of examples:', len(dataset))  # calls __len__

# Access the examples one by one
for i in range(len(dataset)):
    x, y = dataset[i]  # calls __getitem__
    print(f'Example {i}: input {x}, target {y}')
```

```python
train_set = MyDataset(train_X, train_y)
test_set = MyDataset(test_X, test_y)
```

```python
# However, we need to batch our Training Dataset. Why do we need batching?

# Without Batching: If you train your model on one sample at a time, the training can be very slow.
#                   Each update will depend on the noise from individual samples, which may lead to unstable training.

# With Batching: If you use a batch size of 8, for instance, the model processes 8 samples in parallel.
#                It computes the average loss and gradients, and then updates the model weights.
#                This leads to faster training times and more stable convergence.
#                However, the larger the batch size the more memory you will need to train your model.

train_loader = DataLoader(train_set, batch_size=8, shuffle=True)  # new random order in every epoch
test_loader = DataLoader(test_set, batch_size=8)  # no need to shuffle for evaluation
```

Recommended `DataLoader` settings:
- `shuffle=True` for the training data, so that the minibatches are different in every epoch; `shuffle=False` (the default) for validation and test data.
- `batch_size` is a hyperparameter: larger batches make an epoch faster on a GPU but need more memory; powers of 2 such as 32 or 64 are common choices.
- If the number of examples is not divisible by the batch size, the last batch is smaller (below: 100 examples, 12 batches of 8 and one of 4); `drop_last=True` drops it.

```python
for batch in train_loader:
    print(batch[0].shape, batch[1].shape)
```

## A Small Neural Net

Defining the neural network model. In PyTorch, a model is usually defined as a class that inherits from torch.nn.Module. It includes layers as class attributes and defines the forward pass of the network.

The model returns *logits*: one raw score per class, without a softmax. The softmax is applied inside the loss function (see below), which is numerically more stable.

```python
class TinyNN(torch.nn.Module):

    def __init__(self, num_features, num_classes):
        super().__init__()
        self.num_features = num_features
        self.num_classes = num_classes
        self.hidden_size = int(round(num_features / 2))
        self.input_layer = torch.nn.Linear(num_features, self.hidden_size)
        self.hidden_layer = torch.nn.Linear(self.hidden_size, self.hidden_size)
        self.output_layer = torch.nn.Linear(self.hidden_size, num_classes)

    def forward(self, inputs):
        out1 = torch.relu(self.input_layer(inputs))  # without a non-linearity, stacked linear layers stay linear
        out2 = torch.relu(self.hidden_layer(out1))
        logits = self.output_layer(out2)  # raw scores, one per class; no softmax here
        return logits
```

## Let's Train it
The training process involves several steps such as setting the model to training mode, looping over the data loader, computing the loss, performing backpropagation to calculate gradients, and updating the model parameters with an optimizer.

```python
import torch.optim as optim

model = TinyNN(num_features=num_features, num_classes=num_classes)
print(f"Device: {device}")
model.to(device)

# CrossEntropyLoss takes logits and class indices: it applies log_softmax and then the negative log-likelihood loss (NLLLoss).
loss_func = torch.nn.CrossEntropyLoss()
lr = 0.001
optimizer = optim.Adam(model.parameters(), lr=lr)

num_epochs = 10
```

```python
num_batches = len(train_loader)
for epoch in range(1, num_epochs + 1):
    model.train()
    epoch_loss = 0.0
    for inputs, y_true in train_loader:
        inputs, y_true = inputs.to(device), y_true.to(device)
        optimizer.zero_grad()
        y_pred = model(inputs)  # logits, shape (batch_size, num_classes)
        loss = loss_func(y_pred, y_true.squeeze(dim=1))  # targets: shape (batch_size, 1) -> (batch_size,)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    print(f'Epoch [{epoch}/{num_epochs}], average loss: {epoch_loss / num_batches:.4f}')
```

## Simple Evaluation:

After training, it's essential to evaluate the model on a separate test dataset to assess its performance on **unseen** data. Here, we're calculating the accuracy of the model, which is a common metric for classification tasks.

Note that our data are random numbers with random labels, so there is nothing to learn: whatever the model picks up from the training data is noise, and the test accuracy fluctuates around chance level (1/4 for 4 classes). With only 50 test examples, it can easily land well below or above 0.25.

```python
predictions = []
true_labels = []
model.eval()  # without dropout and alike not really necessary
with torch.no_grad():  # disable gradient computation, since it is only needed when backward() is called
    for inputs, y_true in test_loader:
        y_pred = model(inputs.to(device))
        batch_preds = [x.item() for x in torch.argmax(y_pred, dim=-1)]
        predictions.extend(batch_preds)
        true_labels.extend([y.item() for y in y_true.squeeze(dim=1)])
```

```python
from sklearn.metrics import accuracy_score
acc = accuracy_score(true_labels, predictions)
acc
```

## Saving and Loading a Model

Training a model can take a long time, and Colab disconnects idle sessions. Save the learned parameters to a file: `model.state_dict()` is a dictionary from parameter names to tensors. To use them later, create a new model of the same class and load the parameters into it. On Colab, save to your mounted Google Drive, otherwise the file disappears with the runtime.

`torch.save` works for any tensor or dictionary of tensors, not only for model parameters.

```python
torch.save(model.state_dict(), 'tiny_nn.pt')

new_model = TinyNN(num_features=num_features, num_classes=num_classes)  # same architecture, new random weights
new_model.load_state_dict(torch.load('tiny_nn.pt', map_location=device))  # replace them with the saved weights
new_model.to(device)
```

```python
# Both models now compute the same outputs
inputs, _ = next(iter(test_loader))
inputs = inputs.to(device)
model.eval()
new_model.eval()
with torch.no_grad():
    print(torch.allclose(model(inputs), new_model(inputs)))
```
